### Code to query required data from Gaia DR3

```python

from astroquery.gaia import Gaia
import matplotlib.pyplot as plt
from os.path import isfile
```

```python
Gaia.login()
```

```python
# TAP-query for downloading data
RERUN = False
filename = "All_candidates.vot"
if (not isfile(filename)) or RERUN:
    query = """
    SELECT
        bp_rp,
        bp_g,
        g_rp,
        l, 
        b, 
        ra,
        ra_error,
        dec,
        dec_error,
        parallax, 
        parallax_error,
        pmra, 
        pmra_error,
        pmdec, 
        pmdec_error,
        phot_g_mean_mag, 
        teff_gspphot,
        bp_rp,
        radial_velocity,
        radial_velocity_error
    FROM 
        gaiadr3.gaia_source AS gdr3
    WHERE
        parallax > 0.5
        AND (1000./parallax * COS(b*PI()/180)) < 200

    """
    # the selection in the specification "where" is further restricted in the analysis and is significantly wider than the actual selection

    # Perform TAP query
    job = Gaia.launch_job_async(query)
    results = job.get_results()

    # Save to file
    results.write(filename, format="votable", overwrite=True)
    print("Data download complete.")
else:
    print("Data file found. Skipping Gaia query.")
```
