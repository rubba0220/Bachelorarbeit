```python
fig = plt.figure(figsize=(20,10))
ns = jnp.linspace(0,5000,100)
atols = 24000/ns**5
ts = 0.263*ns + 170
plt.plot(atols, ts, label='Dopri5')
ns = jnp.linspace(0, 30000, 100)
atols = 120000/ns**4
ts = 0.01574*ns+7.59
plt.plot(atols, ts, label='simple RK4')
plt.xscale('log')
plt.xlim(1e-5, 1e-12)
plt.ylim(0, 800)
plt.legend()
plt.xlabel('$atol$')
plt.ylabel('$t/\mu s$')
plt.grid()
plt.savefig('atol_vs_t.png')
```
